## <u>Введение</u>

В настоящем проекте необходимо разработать демонстрационную версию системы для поиска изображений по текстовому запросу.

Для демонстрационной версии необходимо обучить модель, которая получит векторное представление изображения, векторное представление текста, а на выходе выдаст число от 0 до 1, т.е. покажет, насколько текст и картинка подходят друг другу.

**Описание данных**

В файле `train_dataset.csv` находится информация, необходимая для обучения: имя файла изображения, идентификатор описания и текст описания. Для одной картинки может быть доступно до 5 описаний. Идентификатор описания имеет формат `<имя файла изображения>#<порядковый номер описания>`.

В папке `train_images` содержатся изображения для тренировки модели.

В файле `CrowdAnnotations.tsv` — данные по соответствию изображения и описания, полученные с помощью краудсорсинга. Номера колонок и соответствующий тип данных:

1. Имя файла изображения.
2. Идентификатор описания.
3. Доля людей, подтвердивших, что описание соответствует изображению.
4. Количество человек, подтвердивших, что описание соответствует изображению.
5. Количество человек, подтвердивших, что описание не соответствует изображению.

В файле `ExpertAnnotations.tsv` содержатся данные по соответствию изображения и описания, полученные в результате опроса экспертов. Номера колонок и соответствующий тип данных:

1. Имя файла изображения.
2. Идентификатор описания.

3, 4, 5 — оценки трёх экспертов.

Эксперты ставят оценки по шкале от 1 до 4, где 1 — изображение и запрос совершенно не соответствуют друг другу, 2 — запрос содержит элементы описания изображения, но в целом запрос тексту не соответствует, 3 — запрос и текст соответствуют с точностью до некоторых деталей, 4 — запрос и текст соответствуют полностью.

В файле `test_queries.csv` находится информация, необходимая для тестирования: идентификатор запроса, текст запроса и релевантное изображение. Для одной картинки может быть доступно до 5 описаний. Идентификатор описания имеет формат `<имя файла изображения>#<порядковый номер описания>`.

В папке `test_images` содержатся изображения для тестирования модели.

### Функция ```word_freq```

````python
def word_freq(data):
    """
    Функция подсчитывает частоту встречаемости слов в текстах 
    и возвращает DataFrame с топ-50 самых частых слов
    
    Параметры:
    ----------
        data : array-like
            Массив или Series с текстами
    
    Возвращает:
    ----------- 
        top_words_df : pandas.DataFrame
            DataFrame с топ-50 слов и их частотами
        word_freq : collections.Counter
            Counter объект со всеми словами и их частотами
    """
    all_words = []
    for text in data:
        words = text.split()
        all_words.extend(words)
    
    # считаем слова и создаем df с топ-50 слов
    word_freq = Counter(all_words)
    return pd.DataFrame(word_freq.most_common(50), 
                               columns=['word', 'frequency']), word_freq
````

### Функция ```plot_word_freq```

````python
def plot_word_freq(data, word_freq):
    """
    Функция строит график топ-50 самых частых слов и выводит статистику
    
    Параметры:
    ----------
        data : pandas.DataFrame
            DataFrame с колонками 'word' и 'frequency'
        word_freq : collections.Counter
            Counter объект со всеми словами и их частотами
    
    Возвращает:
    ----------- 
        None
    """
    plt.figure(figsize=(13, 10))
    bars = plt.barh(data['word'], data['frequency'], 
                    color='lightblue', edgecolor='black', linewidth=0.5)
    
    for i, (word, freq) in enumerate(zip(data['word'], data['frequency'])):
        plt.text(freq + 0.1, i, str(freq), va='center', fontsize=6)
    
    plt.xlabel('Частота встречаемости', fontsize=12)
    plt.ylabel('Слова', fontsize=12)
    plt.title('Топ-50 самых частых слов в текстах', fontsize=12, fontweight='bold', pad=20)
    plt.gca().invert_yaxis()
    plt.grid(axis='x', alpha=0.2, linestyle='--')
    plt.tight_layout()
    plt.show()
    
    # статистика
    print(f"Общая статистика:")
    print(f"Всего уникальных слов: {len(word_freq):,}")
    print(f"Общее количество слов: {sum(word_freq.values()):,}")
    print(f"Средняя длина текста: {sum(word_freq.values()) / len(data):.1f} слов")
    print(f"\nТоп-10 самых частых слов:")
    print(data.head(10).to_string(index=False))
````

### Функция ```plot_data_analysis```

````python
# вспомогательная функция для отрисовки параметров с целью их дальнейшего анализа
# функция была написана ранее в рамках предыдущих проектов, эта версия доработанная
def plot_data_analysis(
    data,
    title: str = 'Изучение параметров',
    title_box: str = 'Диаграмма размаха',
    title_hist: str = 'Распределение',
    x_label: str = 'X',
    y_label: str = 'Y',
    plot_box: bool = True,
    plot_bar: bool = False,
    x = None,
    y = None,
    bins: int = 200,
    bar_labels = None,
    discrete = False,
    figsize = None,
    color: str = 'skyblue',
    show_stats: bool = True,
    label_angle: int = 0,
    log: bool = False
):
    """
    Функция для отрисовки данных с возможностью построения:
        boxplot (диаграмма размаха)
        гистограммы или barplot (столбчатой диаграммы)
    
    Параметры:
    ----------
        data -- данные для анализа (pd.Series, pd.DataFrame или массив)
        title -- общий заголовок
        title_box -- заголовок для boxplot
        title_hist -- заголовок для гистограммы/barplot
        x_label -- подпись оси X
        y_label -- подпись оси Y
        plot_box -- строить ли boxplot (по умолчанию True)
        plot_bar -- использовать ли barplot вместо гистограммы (по умолчанию False)
        x -- данные для оси X (если нужен barplot с внешними данными)
        y -- данные для оси Y (если нужен barplot с внешними данными)
        bins -- количество бинов для гистограммы
        bar_labels -- подписи для barplot
        discrete -- флаг для дискретных данных
        figsize -- размер графика
        color -- основной цвет графиков
        show_stats -- показывать ли статистику (по умолчанию True)
    """
    # проверка и преобразование типов ходных данных
    if not isinstance(data, (pd.Series, pd.DataFrame, np.ndarray, list)):
        raise TypeError("Данные должны быть типа pd.Series, pd.DataFrame, np.ndarray или list")
    
    if isinstance(data, (pd.DataFrame, np.ndarray, list)):
        data = pd.Series(data)
    
    # настройка размера графика
    if figsize is None:
        figsize = (13, 8) if plot_box else (13, 6)
    
    # создание subplots
    if plot_box:
        fig, (ax_box, ax_main) = plt.subplots(2, 1, figsize=figsize, 
                                             gridspec_kw={'height_ratios': [1, 2]})
        # усы
        ax_box.boxplot(data, vert=False, patch_artist=True,
                      boxprops=dict(facecolor='lightblue'))
        ax_box.set_title(title_box, pad=10)
        ax_box.set_xlabel(x_label)
        ax_box.grid(axis='x', linestyle='--', alpha=0.7)
        ax_box.set_yticks([])
    else:
        fig, ax_main = plt.subplots(1, 1, figsize=figsize)
    
    # выбор типа основного графика
    if plot_bar:
        if x is None or y is None:
            if discrete:
                value_counts = data.value_counts().sort_index()
                x = value_counts.index
                y = value_counts.values
                bar_labels = x if bar_labels is None else bar_labels
            else:
                raise ValueError("Для barplot нужно указать x и y или использовать discrete=True")
        
        ax_main.bar(x=x, height=y, color=color, edgecolor='black', alpha=0.8)
        ax_main.set_title(title_hist, pad=10)
        
        # установка подписей для дискретных данных
        if bar_labels is not None:
            ax_main.set_xticks(x if discrete else np.arange(len(x)))
            ax_main.set_xticklabels(bar_labels, rotation=label_angle, ha='right')
    else:
        # гистограмма
        ax_main.hist(data, bins=bins, color=color, edgecolor='black', alpha=0.8, log=log)
        ax_main.set_title(title_hist, pad=10)
    
    # подписи к осям и сетка
    ax_main.set_xlabel(x_label)
    ax_main.set_ylabel(y_label)
    ax_main.grid(axis='both', linestyle='--', alpha=0.5)
    
    fig.suptitle(title, fontsize=14, y=1.02)
    plt.tight_layout()
    plt.show()
    
    # вывод статистики
    if show_stats:
        display(data.describe())
````

### Функция ```bar_plot```

````python
def bar_plot(data, 
             ax, 
             title,
             width = 0.6):
    """
    Упрощенная версия plot_data_analysis, просто построим бары

    Параметры:
    ----------
        data -- данные
        ax -- оси matplotlib
        title -- заголовок
        width -- ширина баров (по умолчанию 0.6)
    """
    value_counts = data.value_counts().sort_index()
    ax.bar(x=value_counts.index, height=value_counts.values, 
           color='skyblue', width=width, edgecolor='black', alpha=0.8)
    ax.set_title(title)
    ax.set_xlabel('Оценка')
    ax.set_ylabel('Количество')
    ax.grid(axis='both', linestyle='--', alpha=0.5)
````

### Функция ```get_final_target```

````python
def get_final_target(data,
                    expert_w=0.9,
                    crowd_w=0.1):
    """
    Функция для объединения признаков экспертов и краудсорсинга
    
    Параметры:
    ----------
        data -- данные
        expert_w -- вес оценки экспертов (по умолчанию 0.9)
        crowd_w -- вес оценки краудсорсинга (по умолчанию 0.1)
    
    Возвращает:
    ----------- 
        Итоговую оценку
    """
    expert = data['target_expert']
    crowd = data['frac']
    
    # если есть обе оценки -- объединим их
    if not pd.isna(expert) and not pd.isna(crowd):
        return expert_w * expert + crowd_w * crowd
    
    # если есть только экспертная оценка
    elif not pd.isna(expert):
        return expert
    
    # если есть только краудсорсинг оценка
    elif not pd.isna(crowd):
        return crowd
    
    # если оценок вообще нет
    else:
        return np.nan
````

### Функция ```plot_isna```

````python
# функция для отображения количества пропусков в процентном соотношении от общего числа записей
def plot_isna(data):
    try:
        # вычисляем процент пропусков
        missing_data = data.isna().mean() * 100
        # фильтруем столбцы с пропусками и сортируем
        missing_data = missing_data[missing_data > 0].sort_values(ascending=True)
        
        if missing_data.empty:
            print('Пропуски в данных отсутствуют.')
            return
        
        # создаем горизонтальную столбчатую диаграмму
        fig, ax = plt.subplots(figsize=(13, 6))
        ax.barh(missing_data.index, missing_data, color='skyblue', edgecolor='black')
        
        # настраиваем подписи и заголовок
        ax.set_title('% пропусков в данных по столбцам\n(от общего количества записей в данных)', fontsize=14)
        ax.set_xlabel('Пропуски в данных, (%)', fontsize=12)
        ax.set_ylabel('Столбцы', fontsize=12)
        
        # настраиваем шрифт для меток на осях
        ax.tick_params(axis='both', labelsize=12)
        
        # добавляем сетку
        ax.grid(True, axis='x', linestyle='--', alpha=0.7)
        
        # строим график
        plt.show()
        print("Доля пропусков по столбцам:")
        display(missing_data)
        
    except Exception as e:
        print(f'Ошибка: {e}')
````

### Функция ```vectorize_img```

````python
def vectorize_img(df, model, preprocess, image_col='image',
                 save_results=False, save_name=None,
                 images_folder='train_images',
                 load_from_file=False):
    """
    Векторизация изображений из DataFrame
    
    Параметры:
    ----------
        df: DataFrame с данными
        image_col: название колонки с именами файлов изображений
        model: модель для векторизации
        preprocess: препроцессор
        save_results: сохранить фичи в файл
        save_name: имя файла для сохранения
        images_folder: папка с изображениями
        load_from_file: загрузить из файла если существует
    
    Возвращает:
    ----------- 
        features_df: DataFrame с названиями файлов и результатами векторизации
    """
    
    # пытаемся загрузить из файла
    if load_from_file and save_name and os.path.exists(save_name):
        
        with open(save_name, 'rb') as f:
            features_dict = pickle.load(f)
        
        # создаем df с названиями файлов картинок и результатами векторизации
        img_names = list(features_dict.keys())
        features_list = [features_dict[img_name] for img_name in img_names]
        feature_columns = [f'img_feature_{i}' for i in range(len(features_list[0]))]
        
        features_df = pd.DataFrame(features_list, columns=feature_columns)
        features_df.insert(0, image_col, img_names)
        
        print(f"Загружено {len(features_dict)} изображений")
        print(f"Размер датасета: {features_df.shape}")
        
        return features_df
    
    # если файл не открылся, делаем векторизацию заново
    features_dict = {}
    failed_images = []
    
    # берем уникальные изображения из указанной колонки
    image_names = df[image_col].unique()

    # векторизуем изображения
    for img_name in tqdm(image_names, desc="Векторизация изображений"):
        img_path = os.path.join(images_folder, img_name)
        
        try:
            with Image.open(img_path) as img:
                img = img.convert('RGB')
            
            image_tensor = preprocess(img).unsqueeze(0).to(device)
            
            with torch.no_grad():
                features = model(image_tensor)
                features = features.cpu().numpy().flatten()
            
            features_dict[img_name] = features
            
        except Exception as e:
            print(f"Ошибка обработки изображения {img_name}: {e}")
            failed_images.append(img_name)
            continue
    
    print(f"Изображений векторизовано: {len(features_dict)}")
    print(f"Изображений с ошибками: {len(failed_images)}")
    
    if save_results and save_name:
        with open(save_name, 'wb') as f:
            pickle.dump(features_dict, f)
        print(f"Векторизованные изображения сохранены: {save_name}")
    
    # создаем результирующий df
    img_names = list(features_dict.keys())
    features_list = [features_dict[img_name] for img_name in img_names]
    feature_columns = [f'img_feature_{i}' for i in range(len(features_list[0]))]
    
    features_df = pd.DataFrame(features_list, columns=feature_columns)
    features_df.insert(0, image_col, img_names)
    
    print(f"Размерность датасета с векторизованными изображениями: {features_df.shape}")
    
    return features_df
````

### Функция ```vectorize_text```

````python
def vectorize_text(df, tokenizer, model, text_col='query_text',
                   batch_size=100, max_length=512,
                   save_results=False, save_name=None,
                   load_from_file=False):
    """
    Векторизация текстов из DataFrame через BERT батчами
    
    Параметры:
    ----------
        df: DataFrame с данными
        text_col: название колонки с текстами
        tokenizer: BERT токенизатор
        model: BERT модель
        batch_size: размер батча
        max_length: максимальная длина (если None - определит автоматически)
        save_results: сохранить фичи в файл
        save_name: имя файла для сохранения
        load_from_file: загрузить из файла если существует
    
    Возвращает:
    ----------- 
        features_df: DataFrame с текстами и результатами векторизации
    """
    
    # получаем устройство из модели
    device = next(model.parameters()).device
    
    # пробуем загрузить из файла
    if load_from_file and save_name and os.path.exists(save_name):
        
        features = np.load(save_name)
        
        # создаем df только с текстами и результатом векторизации
        texts = df[text_col].unique()
        feature_columns = [f'text_feature_{i}' for i in range(features.shape[1])]
        
        features_df = pd.DataFrame(features, columns=feature_columns)
        # добавляем тексты в первый столбец
        features_df.insert(0, text_col, texts)  
        
        print(f"Загружено {features.shape[0]} текстовых векторов")
        print(f"Размер датасета: {features_df.shape}")
        
        return features_df
    
    # если не загружается файл, то векторизуем заново
    # берем тексты из указанной колонки
    texts = df[text_col].unique()
    
    # токенизация
    tokenized = [tokenizer.encode(text, add_special_tokens=True) for text in texts]
    
    # определяем максимальную длину автоматом, если иное не указали
    if max_length is None:
        max_length = max(len(i) for i in tokenized)
    
    # паддинг и attention mask
    padded = np.array([i + [0] * (max_length - len(i)) for i in tokenized])
    attention_mask = np.where(padded != 0, 1, 0)
    
    embeddings = []

    # делаем векторизацию
    for i in tqdm(range(0, len(padded), batch_size), desc="BERT векторизация"):
        batch_end = min(i + batch_size, len(padded))
        batch = torch.LongTensor(padded[i:batch_end]).to(device)
        attention_mask_batch = torch.LongTensor(attention_mask[i:batch_end]).to(device)
        
        with torch.no_grad():
            batch_embeddings = model(batch, attention_mask=attention_mask_batch)
        
        embeddings.append(batch_embeddings[0][:, 0, :].cpu().numpy())
    
    features = np.concatenate(embeddings)
    print(f"Текстов векторизировано: {features.shape[0]}")
    print(f"Размерность векторизованных текстов: {features.shape[1]}")
    
    # сохраняем результаты, если указано во флаге
    if save_results and save_name:
        np.save(save_name, features)
        print(f"Векторизованные тексты сохранены: {save_name}")
    
    # создаем результирующий df
    feature_columns = [f'text_feature_{i}' for i in range(features.shape[1])]
    features_df = pd.DataFrame(features, columns=feature_columns)
    features_df.insert(0, text_col, texts)
    
    print(f"Размерность датасета с векторизованными текстами: {features_df.shape}")
    
    return features_df
````

### Функция ```search_and_display```

````python
# функция для поиска и отображения наиболее похожих изображений по текстовому запросу
def search_and_display(model, query_text, test_df, tokenizer, bert_model,
                      banned_words=None, images_folder='test_images', top_k=4, device='cuda'):
    model.eval()

    # проверяем на требования законодательства
    if banned_words and any(word in query_text.lower() for word in banned_words):
        print(f"Поисковый запрос: '{query_text}'")
        print('This image is unavailable in your country in compliance with local laws.')
        return None

    print(f"Поисковый запрос: '{query_text}'")

    ground_truth_rows = test_df[test_df['query_text'] == query_text]
    ground_truth_img = ground_truth_rows.iloc[0]['image'] if not ground_truth_rows.empty else None

    # векторизуем запрос
    text_vector = get_text_vector(query_text, tokenizer, bert_model, device).flatten()
    # нормализуем вектор текста
    text_vector = normalize(text_vector.reshape(1, -1), norm='l2').flatten()
    
    unique_images = test_df['image'].unique()
    img_vectors = []
    for img_name in unique_images:
        row = test_df[test_df['image'] == img_name].iloc[0]
        img_feats = [row[f'img_feature_{i}'] for i in range(512)]
        img_vectors.append(img_feats)
    img_vectors = np.array(img_vectors, dtype=np.float32)

    # нормализуем векторы изображения
    img_vectors = normalize(img_vectors, norm='l2')

    # объединяем вектора
    text_matrix = np.tile(text_vector, (len(unique_images), 1))
    combined_matrix = np.concatenate([text_matrix, img_vectors], axis=1)

    combined_tensor = torch.FloatTensor(combined_matrix).to(device)
    with torch.no_grad():
        scores = model(combined_tensor).cpu().numpy().flatten()

    predictions_dict = {}
    for img_name, score in zip(unique_images, scores):
        example_text = test_df[test_df['image'] == img_name].iloc[0]['query_text']
        predictions_dict[img_name] = (score, example_text)

    predictions = [(score, img_name, text) for img_name, (score, text) in predictions_dict.items()]
    predictions.sort(reverse=True, key=lambda x: x[0])
    top_predictions = predictions[:top_k]

    total_cols = top_k + (1 if ground_truth_img else 0)
    fig = plt.figure(figsize=(4 * total_cols, 6))
    col_idx = 0

    if ground_truth_img:
        ax_gt = fig.add_subplot(1, total_cols, col_idx + 1)
        gt_path = os.path.join(images_folder, ground_truth_img)
        try:
            img_gt = Image.open(gt_path)
            ax_gt.imshow(img_gt)
            ax_gt.set_title(f"Корректное изображение: \n{ground_truth_img}", fontsize=11, color='green')
            ax_gt.axis('off')
        except Exception as e:
            ax_gt.text(0.5, 0.5, "Ошибка!", ha='center', va='center', color='red')
            ax_gt.axis('off')
        col_idx += 1

    for i, (score, img_name, img_query_text) in enumerate(top_predictions):
        ax = fig.add_subplot(1, total_cols, col_idx + i + 1)
        img_path = os.path.join(images_folder, img_name)
        try:
            img = Image.open(img_path)
            ax.imshow(img)
            rank = i + 1
            title = f"#{rank} | {score:.3f}\n{img_name}"
            color = 'green' if ground_truth_img and img_name == ground_truth_img else 'black'
            ax.set_title(title, fontsize=10, color=color)
            ax.axis('off')
        except Exception as e:
            ax.text(0.5, 0.5, "Ошибка!", ha='center', va='center', color='red')
            ax.axis('off')

    plt.suptitle(f"Текстовый запрос: \"{query_text}\"", fontsize=14, y=0.98)
    plt.tight_layout()
    plt.show()

    print("\nРезультаты:")
    if ground_truth_img:
        print(f"Корректное изображение: {ground_truth_img}")
    print(f"ТОП-{top_k} предсказанных изображений:")
    for i, (score, img_name, img_query_text) in enumerate(top_predictions, 1):
        marker = "Корректное изображение" if ground_truth_img and img_name == ground_truth_img else ""
        print(f"{i}. {img_name} | Метрика: {score:.3f}{marker}")
        print(f"Текстовый запрос: {img_query_text}")

    return top_predictions
````

### Функция ```get_text_vector```

````python
# Функция для векторизации текстового запроса
def get_text_vector(text, 
                    tokenizer, 
                    bert_model, 
                    device, 
                    max_length=512):
    """Векторизация одного текста через BERT"""
    inputs = tokenizer(text, return_tensors='pt', truncation=True, 
                      padding=True, max_length=max_length).to(device)
    with torch.no_grad():
        outputs = bert_model(**inputs)
    return outputs.last_hidden_state[:, 0, :].cpu().numpy()
````

### Класс ```NNet```

````python
# класс модели взял из своего проекта по предсказанию температуры поверхности звезды
class NNet(nn.Module):
    """
    Универсальный класс нейронной сети на базе nn.Sequential с поддержкой BatchNorm.
    
    Класс создаёт полносвязную нейронную сеть с гибкой настройкой архитектуры,
    ф-ций активации, инициализации весов и метода обучения.
    
    Применение:
    ----------
    - Регрессия с произвольной архитектурой
    - Классификация с различными выходными активациями
    - Эксперименты с различными инициализациями и нормализациями
    
    Пример использования:
    net = NNet([10, 64, 32, 1], activation='ReLU', dropout_rate=0.2, 
               use_batchnorm=True)
    """
    
    def __init__(self, n_neurons, activation='ReLU', dropout_rate=0.0, 
                 weight_init='kaiming', bias_init='zeros', output_activation=None,
                 activation_params=None, use_batchnorm=False, bn_momentum=0.1):
        """
        Инициализация универсальной нейронной сети.
        
        Параметры:
        ----------
        n_neurons -- список целых чисел, определяющий архитектуру сети
                     (первый элемент - размер входа, последний - выход)
        activation -- строка, тип активационной функции для скрытых слоёв
                      возможные варианты: 'ReLU', 'LeakyReLU', 'ELU', 'SiLU', 
                      'Tanh', 'Sigmoid'
        dropout_rate -- вещественное число, вероятность dropout (0.0 = без dropout)
        weight_init -- строка, метод инициализации весов
                       варианты: 'normal', 'xavier', 'kaiming'
        bias_init -- строка, метод инициализации смещений
                     варианты: 'normal', 'zeros', 'constant'
        output_activation -- строка или None, активация для выходного слоя
                             варианты: 'Sigmoid', 'Softmax', None
        activation_params -- словарь или None, параметры активаций
                             например: {'alpha': 0.01} для LeakyReLU
        use_batchnorm -- булево, использовать ли BatchNorm1d после линейных слоёв
        bn_momentum -- вещественное число, момент для Batch Normalization (по умолчанию 0.1)
        """
        # базовая инициализация родительского класса
        super(NNet, self).__init__()
        
        # проверка и установка параметров активации по умолчанию
        if activation_params is None:
            activation_params = {}
        
        # список для хранения всех слоёв сети
        layers = []
        
        # основной цикл построения архитектуры
        for i in range(len(n_neurons) - 1):
            # добавление линейного слоя
            layers.append(nn.Linear(n_neurons[i], n_neurons[i+1]))
            
            # BatchNorm после линейного слоя для скрытых слоёв (не для последнего)
            if i < len(n_neurons) - 2 and use_batchnorm:
                layers.append(nn.BatchNorm1d(n_neurons[i+1], momentum=bn_momentum))
            
            # активация и dropout для скрытых слоёв (не для последнего)
            if i < len(n_neurons) - 2:
                # выбор активационной функции
                if activation == 'ReLU':
                    layers.append(nn.ReLU())
                elif activation == 'LeakyReLU':
                    alpha = activation_params.get('alpha', 0.01)
                    layers.append(nn.LeakyReLU(negative_slope=alpha))
                elif activation == 'ELU':
                    alpha = activation_params.get('alpha', 1.0)
                    layers.append(nn.ELU(alpha=alpha))
                elif activation == 'SiLU':  # Swish активация
                    layers.append(nn.SiLU())
                elif activation == 'Tanh':
                    layers.append(nn.Tanh())
                elif activation == 'Sigmoid':
                    layers.append(nn.Sigmoid())
                else:
                    # обработка неподдерживаемой активации
                    raise ValueError(f"Неподдерживаемая активация: {activation}")
                
                # добавление dropout для скрытых слоёв при необходимости
                if dropout_rate > 0:
                    layers.append(nn.Dropout(p=dropout_rate))
            else:
                # активация выходного слоя (если указана)
                if output_activation == 'Sigmoid':
                    layers.append(nn.Sigmoid())
                elif output_activation == 'Softmax':
                    layers.append(nn.Softmax(dim=1))
                # для регрессии обычно без выходной активации
        
        # создаем модель из всех слоёв
        self.model = nn.Sequential(*layers)
        
        # инициализация весов с учётом ф-ции активации
        self._init_weights(weight_init, bias_init, activation)
        
        # сохранение параметров
        self.use_batchnorm = use_batchnorm
        self.dropout_rate = dropout_rate
        self.activation = activation
        
        # Автоматическое определение device и перемещение модели
        self.device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
        self.to(self.device)
    
    def _init_weights(self, weight_init, bias_init, activation):
        """
        Инициализация весов и смещений с учётом типа нелинейности.
        """
        def init_fn(layer):
            # применение только к линейным слоям
            if isinstance(layer, nn.Linear):
                # выбор нелинейности для kaiming инициализации
                nonlinearity = 'relu'  # значение по умолчанию
                if activation in ['Tanh', 'Sigmoid']:
                    nonlinearity = 'tanh' if activation == 'Tanh' else 'sigmoid'
                elif activation == 'LeakyReLU':
                    nonlinearity = 'leaky_relu'
                
                # инициализация весов
                if weight_init == 'normal':
                    nn.init.normal_(layer.weight, mean=0.0, std=0.02)
                elif weight_init == 'xavier':
                    nn.init.xavier_uniform_(layer.weight)
                elif weight_init == 'kaiming':
                    nn.init.kaiming_normal_(layer.weight, nonlinearity=nonlinearity)
                else:
                    raise ValueError(f"Неподдерживаемая инициализация весов: {weight_init}")
                
                # инициализация смещений
                if bias_init == 'normal':
                    nn.init.normal_(layer.bias, mean=0.0, std=0.02)
                elif bias_init == 'zeros':
                    nn.init.zeros_(layer.bias)
                elif bias_init == 'constant':
                    nn.init.constant_(layer.bias, 0.1)
                else:
                    raise ValueError(f"Неподдерживаемая инициализация смещений: {bias_init}")
        
        self.model.apply(init_fn)
    
    def forward(self, x):
        """
        Прямой проход через сеть.
        """
        # Автоматическое перемещение входных данных на правильное устройство
        if not x.is_cuda and self.device.type == 'cuda':
            x = x.to(self.device)
        return self.model(x).flatten()
    
    def train_model(self, X_train, y_train, X_test, y_test, 
                learning_rate=1e-3, num_epochs=15000, batch_size=32,
                loss_fn=None, optimizer_type='Adam', print_every=500):
        """
        Обучение нейронной сети с автоматическим учетом device.
        """
        # Автоматическое перемещение данных на правильное устройство
        X_train = X_train.to(self.device)
        y_train = y_train.to(self.device)
        X_test = X_test.to(self.device)
        y_test = y_test.to(self.device)
        
        # настройка функции потерь
        if loss_fn is None:
            loss_fn = nn.MSELoss()

        # выбор и настройка оптимизатора
        if optimizer_type == 'Adam':
            optimizer = torch.optim.Adam(self.parameters(), lr=learning_rate)
        elif optimizer_type == 'SGD':
            optimizer = torch.optim.SGD(self.parameters(), lr=learning_rate)
        elif optimizer_type == 'RMSprop':
            optimizer = torch.optim.RMSprop(self.parameters(), lr=learning_rate)
        else:
            raise ValueError(f"Неподдерживаемый оптимизатор: {optimizer_type}")

        # список для хранения истории обучения
        history_list = []

        # определение режима обучения
        use_full_batch = (batch_size <= 0)
        if use_full_batch:
            num_batches = 1
        else:
            train_dataset = TensorDataset(X_train, y_train)
            train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
            num_batches = len(train_loader)

        # основной цикл обучения
        for epoch in range(num_epochs):
            self.train()
            total_loss = 0

            if use_full_batch:
                optimizer.zero_grad()
                preds = self.forward(X_train)
                loss_value = loss_fn(preds, y_train.flatten())
                loss_value.backward()
                optimizer.step()
                total_loss = loss_value.item()
            else:
                for batch_X, batch_y in train_loader:
                    optimizer.zero_grad()
                    preds = self.forward(batch_X)
                    loss_value = loss_fn(preds, batch_y.flatten())
                    loss_value.backward()
                    optimizer.step()
                    total_loss += loss_value.item()

            # вычисление метрик
            self.eval()
            with torch.no_grad():
                train_preds = self.forward(X_train).flatten()
                train_rmse = torch.sqrt(loss_fn(train_preds, y_train.flatten()))

                test_preds = self.forward(X_test).flatten()
                test_rmse = torch.sqrt(loss_fn(test_preds, y_test.flatten()))

            avg_loss = total_loss / num_batches

            history_list.append({
                'epoch': epoch + 1,
                'loss': avg_loss,
                'rmse_train': train_rmse.item(),
                'rmse_test': test_rmse.item()
            })

            if epoch % print_every == 0 or epoch == num_epochs - 1:
                bn_status = "BN" if self.use_batchnorm else "No BN"
                device_status = f"[{self.device.type.upper()}, {bn_status}]"
                print(f'Эпоха: {epoch:4d}, loss: {avg_loss:.6f}, '
                      f'RMSE (train): {train_rmse:.4f}, RMSE (test): {test_rmse:.4f} {device_status}')

        history_df = pd.DataFrame(history_list)
        self.eval()
        return history_df
    
    def predict(self, X, return_numpy=False):
        """
        Получение предсказаний для новых данных.
        
        Параметры:
        ----------
        X -- входные данные для предсказания (тензор)
        return_numpy -- если True, возвращает numpy array вместо тензора
        
        Возвращает:
        ----------
        предсказания в виде тензора или numpy array
        """
        self.eval()
        
        if not X.is_cuda and self.device.type == 'cuda':
            X = X.to(self.device)
                
        with torch.no_grad():
            result = self.forward(X).flatten()
            if return_numpy:
                return result.cpu().numpy()
            return result
    
    def get_metrics(self, X, y, loss_fn=None):
        """
        Вычисление метрик качества модели.
        """
        if loss_fn is None:
            loss_fn = nn.MSELoss()
        
        X = X.to(self.device)
        y = y.to(self.device)
        
        self.eval()
        with torch.no_grad():
            preds = self.predict(X)
            mse = loss_fn(preds, y.flatten())
            rmse = torch.sqrt(mse)
            return {'MSE': mse.item(), 'RMSE': rmse.item()}
    
    def get_architecture_info(self):
        """
        Вывод информации об архитектуре сети.
        """
        linear_layers = self.model[::2] if len(self.model) > 1 else [self.model[0]]
        info = f"Архитектура: {' -> '.join(map(str, linear_layers))} (линейный слои)"
        
        if self.use_batchnorm:
            info += f"\nBatchNorm: Да (момент = {0.1})"
        
        if self.dropout_rate > 0:
            info += f"\nDropout: {self.dropout_rate}"
        
        info += f"\nФ-ция активации: {self.activation}"
        info += f"\nУстройство: {self.device}"
        
        return info
    
    def get_device_info(self):
        """
        Дополнительный метод для получения информации об устройстве.
        """
        return f"Модель на устройстве: {self.device}"
````

### Экспертные оценки

**Комментарий**: рассмотрим для начала экспертные оценки. 

В файле с экспертными мнениями для каждой пары изображение-текст имеются оценки от трёх специалистов. Для решения задачи эти оценки необходимо агрегировать, т.е. превратить в одну.
Предлагается реализовать т.н. "мягкое голосование". В этом случае оценку эксперта можно интерпертировать, как его уверенность в том, что текст соответствует изображению.

- Оценки 1 и 2 -> Не соответствует (0)

- Оценки 3 и 4 -> Соответствует (1)

Преимещуства такого подход:
 - Мы сохраняем и не выбрасываем данные;
 - Создаем плавную целевую переменную от 0 до 1.
 - Учитываем мнение всех экспертов, а не только большинства.

Для каждой пары "текст-изображение" рассчитаем долю экспертов, проголосовавших за "соответствует".

````python
# сконвертируем экспертные оценки в бинарные: соответствует/не соответствует
# а затем сразу посчитаем target_expert как среднее бинарных оценок
expert_annotations['target_expert'] = (
    expert_annotations[['expert_1', 'expert_2', 'expert_3']]
    .map(lambda x: 1 if x in [3, 4] else (0 if x in [1, 2] else np.nan))
    .mean(axis=1)
    )
````

## <u>Проверка данных</u>

**Комментарий**: в некоторых странах, где работает компания, для которой выполняется настоящий проект, действуют ограничения по обработке изображений: поисковым сервисам и сервисам, предоставляющим возможность поиска, запрещено без разрешения родителей или законных представителей предоставлять любую информацию, в том числе, но не исключительно тексты, изображения, видео и аудио, содержащие описание, изображение или запись голоса детей. Ребёнком считается любой человек, не достигший 16 лет.

Необходимо реализовать проверку -- при попытке посмотреть изображения, запрещённые законодательством, вместо картинок показывается дисклеймер:

**This image is unavailable in your country in compliance with local laws**.

Все изображения, которые нарушают данный закон, удалим из обучающей выборки.

````python
# составим список ключевых слов для проверки
restricted_words = [
    'baby', 'babies', 'newborn', 'infant', 'toddler', 
    'child', 'children', 'kid', 'kids', 'teenager', 'teen',
    'adolescent', 'youth', 'youngster', 'minor', 'juvenile',
    'preschooler', 'schoolchild', 'schoolboy', 'schoolgirl',
    'under 16', 'under sixteen', 'minor age', 'underage',
    'son', 'daughter', 'boy', 'boys', 'girls', 
    'brother', 'sister', 'sibling',
    'grandson', 'granddaughter', 
    'nephew', 'niece', 'childhood',
    'preschool', 'kindergarten', 'nursery',
    'pupil', 'classmate', 'homework', 'schoolwork', 'classroom',
    'bedtime story', 'growing up', 'underage'
]
````

````python
# найдем для начала описания, которые содержат слова из списка restricted_words
pattern = r'\b(' + '|'.join(restricted_words) + r')\b'
mask_bad_comments = train_dataset['query_text'].str.contains(pattern, case=False, na=False)

bad_comments = train_dataset[mask_bad_comments]
print(f"Найдено неподобающих описаний изображений: {len(bad_comments)}")

# для найденных описаний берем query_id и извлекаем из них имена картинок
bad_images = set()
for query_id in bad_comments['query_id']:
    image_name = query_id.split('#')[0]
    bad_images.add(image_name)

print(f"Найдено неподобающих изображений: {len(bad_images)}")
````

````python
# посмотрим на 8 случайных картинок
print("\nСлучайные 8 неподобающих изображений для удаления:")
bad_images_list = list(bad_images)
random_bad_images = random.sample(bad_images_list, min(8, len(bad_images_list)))

fig, axes = plt.subplots(2, 4, figsize=(15, 8))
axes = axes.ravel()

for i, img_name in enumerate(random_bad_images):
    try:
        # путь к картинкам
        img_path = os.path.join(PATH_TO_DATA, "train_images", img_name)
        img = Image.open(img_path)
        axes[i].imshow(img)
        
        # приводим пример неподобающего описания для этой картинки
        bad_text_sample = bad_comments[bad_comments['query_id'].str.startswith(img_name)]['query_text'].iloc[0]
        axes[i].set_title(f"{img_name}\n{bad_text_sample[:30]}...", fontsize=6)
        axes[i].axis('off')
    except Exception as e:
        axes[i].text(0.5, 0.5, f"Не удалось\nзагрузить\n{img_name}", 
                    ha='center', va='center', transform=axes[i].transAxes, fontsize=8)
        axes[i].set_title(f"Ошибка загрузки изображения", fontsize=8)
        axes[i].axis('off')

plt.tight_layout()
plt.show()

for i, img_name in enumerate(random_bad_images, 1):
    print(f"{i}. Изображение: {img_name}")
    img_data = bad_comments[bad_comments['query_id'].str.startswith(img_name)]
    unique_comments = img_data.drop_duplicates(subset=['query_text'])

    for j, (idx, row) in enumerate(unique_comments.iterrows(), 1):
        print(f"   Текстовое описание: {row['query_text']}")
````

## <u>Обучение модели предсказания соответствия</u>

**Комментарий**: выполним теперь разбиение на тренировочную и валидационную выборки, используя ```GroupShuffleSplit```. При формировании выборок также выполним нормализацию векторов изображений и текстов.

В качестве модели будем использовать полносвязную нейронную сеть для решения задачи регрессии target в диапазоне $[0;1]$. На выходе 1 нейрон.

Для такой задачи выбрана метрика $RMSE$ (Root Mean Squared Error -- СКО), т.к.: 
  - Классическая метрика для регрессионных задач;
  - Метрика измеряется в тех же единицах, что и целевая переменная, что упрощает анализ качества модели;
  - Устойчива к выбросам;

````python
# выполняем разбиение
gss = GroupShuffleSplit(n_splits=1, train_size=(1-TEST_SIZE), random_state=RANDOM_STATE)

train_indices, valid_indices = next(gss.split(
    X=final_dataset_with_features.drop(columns=['target']), 
    y=final_dataset_with_features['target'], 
    groups=final_dataset_with_features['image']
))

# получаем тренировочный и валидационный датасеты
train_df = final_dataset_with_features.iloc[train_indices]
valid_df = final_dataset_with_features.iloc[valid_indices]

print(f"Размерность тренировочной выборки: {train_df.shape}")
print(f"Размерность валидационной выборки: {valid_df.shape}")
print(f"Уникальных изображений в тренировочной выборке: {train_df['image'].nunique()}")
print(f"Уникальных изображений в валидационной выборке: {valid_df['image'].nunique()}")
````

````python
# формируем тренировочную выборку
X_train = train_df.drop(columns=['target', 'image'])
y_train = train_df['target']

# формируем валидационную выборку
X_valid = valid_df.drop(columns=['target', 'image'])
y_valid = valid_df['target']
````

````python
# нормализуем данные
X_train_np = X_train.values.astype(np.float32)
X_val_np = X_valid.values.astype(np.float32)
X_train_norm = normalize(X_train_np, norm='l2', axis=1)
X_val_norm = normalize(X_val_np, norm='l2', axis=1)
````

````python
# преобразуем в торч-тензоры
X_train_t = torch.FloatTensor(X_train_norm)
X_val_t = torch.FloatTensor(X_val_norm)
y_train_t = torch.FloatTensor(y_train.values)
y_val_t = torch.FloatTensor(y_valid.values)

print(f"Размерность данных:")
print(f"X_train: {X_train_t.shape}, y_train: {y_train_t.shape}")
print(f"X_val: {X_val_t.shape}, y_val: {y_val_t.shape}")
````

````python
# создаем нейросеть на базее класса NNet
input_size = X_train_t.shape[1]
n_neurons = [input_size, 512, 256, 128, 64, 32, 1]
nn_model = NNet(
    n_neurons=n_neurons,
    activation='LeakyReLU',
    dropout_rate=0.4,
    use_batchnorm=True,
    weight_init='kaiming'
)
````

````python
# выполняем обучение модели
history = nn_model.train_model(
    X_train=X_train_t,
    y_train=y_train_t, 
    X_test=X_val_t,
    y_test=y_val_t,
    learning_rate=1e-3,
    num_epochs=50,
    batch_size=64,
    optimizer_type='Adam',
    print_every=10
)
````

````python
# выводим итоговую метрику
print("Оценка модели:")
metrics = nn_model.get_metrics(X_val_t, y_val_t)
print(f"MSE: {metrics['MSE']:.6f}")
print(f"RMSE: {metrics['RMSE']:.6f}")
````

````python
# построим графики с результатами обучения модели
plt.figure(figsize=(13, 4))

# график функции потерь в зависимости от эпохи при обучении
plt.subplot(1, 3, 1)
plt.plot(history['epoch'], history['loss'], label='Знач. ф-ции потерь')
plt.xlabel('Эпоха')
plt.ylabel('Знач. ф-ции потерь')
plt.title('Значение функции потерь при обучении')
plt.legend()
plt.grid(True)

# график метрики
plt.subplot(1, 3, 2)
plt.plot(history['epoch'], history['rmse_train'], label='RMSE (train)')
plt.plot(history['epoch'], history['rmse_test'], label='RMSE (val)')
plt.xlabel('Эпоха')
plt.ylabel('RMSE')
plt.title('Метрика')
plt.legend()
plt.grid(True)

# предсказания vs фактические оценки соответствия изображений
plt.subplot(1, 3, 3)
with torch.no_grad():
    predictions = nn_model.predict(X_val_t, return_numpy=True)
    
plt.scatter(y_valid, predictions, alpha=0.5)
plt.plot([y_valid.min(), y_valid.max()], [y_valid.min(), y_valid.max()], 'r--', linewidth=2)
plt.xlabel('Истинные значения')
plt.ylabel('Предсказанные значения')
plt.title(f'Предсказанные vs Истинные значения\nRMSE: {metrics["RMSE"]:.4f}')
plt.grid(True)

plt.tight_layout()
plt.show()
````

**Промежуточный вывод**: Видим, что модель начинает быстро переобучаться. Точность модели по результирующей метрике также оставляет желать лучшего.
(тестировались разные архитектуры сети с большим и меньшим количеством слоёв).
